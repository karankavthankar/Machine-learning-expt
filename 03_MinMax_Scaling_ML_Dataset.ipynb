# Min-Max Scaling of ML Dataset

The original script used Google Colab upload controls. This Anaconda version reads `ML.csv` from the notebook folder if present; otherwise it creates sample data with the same expected columns.

## Anaconda / Jupyter Notebook
Run the cells from top to bottom. If a required package is missing, run `%pip install package-name` in a notebook cell, then restart the kernel if needed.


```python
import pandas as pd

try:
    df = pd.read_csv("ML.csv")
    print("Loaded ML.csv")
except FileNotFoundError:
    df = pd.DataFrame({
        "AGE": [20, 25, 30, 35, 40],
        "ANNUAL_INCOME": [200000, 300000, 450000, 600000, 800000],
        "SALARY": [15000, 22000, 30000, 42000, 55000]
    })
    print("ML.csv not found. Using sample data.")

print("Original data:")
display(df)
```

```python
cols = ["AGE", "ANNUAL_INCOME", "SALARY"]
print("Minimum values:")
print(df[cols].min())
print("\nMaximum values:")
print(df[cols].max())

scaled_df = df.copy()
for col in cols:
    minimum = scaled_df[col].min()
    maximum = scaled_df[col].max()
    if maximum != minimum:
        scaled_df[col] = (scaled_df[col] - minimum) / (maximum - minimum)
    else:
        scaled_df[col] = 0.0

print("\nAfter Min-Max scaling:")
display(scaled_df)
scaled_df.to_csv("scaled_dataset.csv", index=False)
print("Saved as scaled_dataset.csv")
```
