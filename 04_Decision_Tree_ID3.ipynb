# Decision Tree using ID3 Algorithm

Implements entropy, information gain, and recursive tree construction on the supplied Play Tennis-style dataset.

```python
import pandas as pd
import numpy as np

data = {
    "Outlook": ["Sunny","Sunny","Overcast","Rain","Rain","Rain","Overcast","Sunny","Sunny","Rain","Sunny","Overcast","Overcast","Rain"],
    "Temperature": ["Hot","Hot","Hot","Mild","Cool","Cool","Mild","Mild","Hot","Mild","Cool","Mild","Cool","Mild"],
    "Humidity": ["High","High","High","High","Normal","Normal","Normal","High","Normal","Normal","High","Normal","Normal","High"],
    "Windy": [False,True,False,False,False,True,True,False,False,False,True,True,False,True],
    "Play": ["No","No","Yes","Yes","Yes","No","Yes","No","Yes","Yes","Yes","Yes","Yes","No"]
}
df = pd.DataFrame(data)
display(df)
```

```python
def entropy(target):
    _, counts = np.unique(target, return_counts=True)
    probabilities = counts / counts.sum()
    return -np.sum(probabilities * np.log2(probabilities))

def information_gain(data, feature, target):
    total_entropy = entropy(data[target])
    values, counts = np.unique(data[feature], return_counts=True)
    weighted_entropy = 0
    for value, count in zip(values, counts):
        subset = data[data[feature] == value]
        weighted_entropy += (count / counts.sum()) * entropy(subset[target])
    return total_entropy - weighted_entropy

def build_tree(data, target):
    classes, counts = np.unique(data[target], return_counts=True)
    if len(classes) == 1:
        return classes[0]
    features = [col for col in data.columns if col != target]
    if not features:
        return classes[np.argmax(counts)]
    gains = [information_gain(data, feature, target) for feature in features]
    best_feature = features[int(np.argmax(gains))]
    tree = {best_feature: {}}
    majority_class = classes[np.argmax(counts)]
    for value in data[best_feature].unique():
        subset = data[data[best_feature] == value].drop(columns=[best_feature])
        if subset.empty:
            tree[best_feature][value] = majority_class
        else:
            tree[best_feature][value] = build_tree(subset, target)
    return tree

tree = build_tree(df, "Play")
print("Entropy:", round(entropy(df["Play"]), 4))
print("Decision tree:")
print(tree)
```
