# K-Means Clustering (from scratch)

Creates four sample groups, assigns points to the nearest centroid, updates centroids, and plots the final clusters.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X1 = np.random.randn(50, 2) + [2, 2]
X2 = np.random.randn(50, 2) + [8, 3]
X3 = np.random.randn(50, 2) + [4, 8]
X4 = np.random.randn(50, 2) + [9, 9]
X = np.vstack((X1, X2, X3, X4))
```

```python
def euclidean_distance(point, centroid):
    return np.sqrt(np.sum((point - centroid) ** 2))

def kmeans(X, k, max_iters=100, seed=42):
    rng = np.random.default_rng(seed)
    centroids = X[rng.choice(len(X), size=k, replace=False)].copy()

    for _ in range(max_iters):
        labels = []
        for point in X:
            distances = [euclidean_distance(point, c) for c in centroids]
            labels.append(np.argmin(distances))
        labels = np.array(labels)

        new_centroids = centroids.copy()
        for i in range(k):
            if np.any(labels == i):
                new_centroids[i] = X[labels == i].mean(axis=0)

        if np.allclose(centroids, new_centroids):
            centroids = new_centroids
            break
        centroids = new_centroids

    # Reassign points using final centroids
    labels = np.array([
        np.argmin([euclidean_distance(point, c) for c in centroids])
        for point in X
    ])
    return centroids, labels

k = 4
centroids, labels = kmeans(X, k)
print("Centroids:\n", centroids)
```

```python
plt.figure(figsize=(8, 5))
for i in range(k):
    points = X[labels == i]
    plt.scatter(points[:, 0], points[:, 1], label=f"Cluster {i + 1}")
plt.scatter(centroids[:, 0], centroids[:, 1], marker="X", color="black", s=200, label="Centroids")
plt.title(f"K-Means Clustering (K={k})")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```
