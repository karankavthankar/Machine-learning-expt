# Data Cleaning

**Corrected version:** the supplied file mainly performed normalization and transformations, which belong in feature transformation rather than data cleaning. This notebook demonstrates actual cleaning: inspect data, handle missing values, remove duplicates, fix data types, and save the cleaned dataset.

```python
import pandas as pd
import numpy as np

# Sample data so the notebook runs immediately.
# Replace this DataFrame with: pd.read_csv("your_file.csv")
df = pd.DataFrame({
    "Name": ["Amit", "Riya", "Amit", "Neha", None, "Vikram"],
    "Age": [20, 21, 20, None, 22, 200],
    "Salary": [25000, 30000, 25000, 28000, None, 40000],
    "City": ["Kolhapur", "Pune", "Kolhapur", "Mumbai", "Pune", "Nashik"]
})

print("Original data:")
display(df)
print("\nShape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nData types:")
print(df.dtypes)
```

```python
# Remove exact duplicate rows
df = df.drop_duplicates()

# Convert numeric columns safely
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

# Replace clearly invalid ages with missing values
df.loc[(df["Age"] < 0) | (df["Age"] > 100), "Age"] = np.nan

# Fill missing numeric values with their median
for col in ["Age", "Salary"]:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with the most frequent value (mode)
for col in ["Name", "City"]:
    df[col] = df[col].fillna(df[col].mode()[0])

# Remove extra spaces and standardize text
df["Name"] = df["Name"].astype(str).str.strip().str.title()
df["City"] = df["City"].astype(str).str.strip().str.title()

print("Cleaned data:")
display(df)
print("\nMissing values after cleaning:")
print(df.isnull().sum())
print("\nDuplicate rows after cleaning:", df.duplicated().sum())
```

```python
# Save the cleaned data in the notebook's current folder
df.to_csv("cleaned_data.csv", index=False)
print("Cleaning completed. Saved as cleaned_data.csv")
```
