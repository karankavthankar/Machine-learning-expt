# Data Transformation and Feature Engineering

This notebook keeps the supplied operations: Min-Max normalization, log transformation, Z-score normalization, and standardization. A sample dataset is included so it runs without a separate Excel file.

## Anaconda / Jupyter Notebook
Run the cells from top to bottom. If a required package is missing, run `%pip install package-name` in a notebook cell, then restart the kernel if needed.


```python
import pandas as pd
import numpy as np

# Use your own file if available:
# df = pd.read_excel("Id.xlsx")
df = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Age": [18, 22, 25, 30, 35],
    "Salary": [18000, 25000, 32000, 45000, 60000]
})
print("Original data:")
display(df)
```

```python
# Min-Max normalization: values are scaled between 0 and 1
minmax_df = df.copy()
cols = ["Age", "Salary"]
minmax_df[cols] = (minmax_df[cols] - minmax_df[cols].min()) / (minmax_df[cols].max() - minmax_df[cols].min())
print("Min-Max normalized data:")
display(minmax_df)
minmax_df.to_excel("normalized_Id.xlsx", index=False)
```

```python
# Log transformation (values must be positive)
log_df = df.copy()
log_df[cols] = np.log(log_df[cols])
print("Log transformed data:")
display(log_df)
log_df.to_excel("log_output.xlsx", index=False)
```

```python
# Z-score normalization / standardization
zscore_df = df.copy()
zscore_df[cols] = (zscore_df[cols] - zscore_df[cols].mean()) / zscore_df[cols].std()
print("Z-score standardized data:")
display(zscore_df)
zscore_df.to_excel("zscore_output.xlsx", index=False)
print("Transformation completed. Excel outputs saved.")
```
