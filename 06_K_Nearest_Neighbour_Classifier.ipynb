# K-Nearest Neighbour Classifier

Uses scikit-learn's built-in breast cancer dataset, scales features, trains KNN, and evaluates the classifier.

## Anaconda / Jupyter Notebook
Run the cells from top to bottom. If a required package is missing, run `%pip install package-name` in a notebook cell, then restart the kernel if needed.


```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df["target"] = data.target
display(df.iloc[:5, :4])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=0.2, random_state=42, stratify=data.target
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=data.target_names))
print("Sample prediction:", data.target_names[model.predict(X_test[:1])[0]])
```
