# Naive Bayes Classifier – Email Spam Detection

Uses the supplied small sample email dataset, converts text to word counts, trains Multinomial Naive Bayes, and predicts a new email.

## Anaconda / Jupyter Notebook
Run the cells from top to bottom. If a required package is missing, run `%pip install package-name` in a notebook cell, then restart the kernel if needed.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

data = {
    "email": [
        "You have won a free vacation",
        "Click here to claim your cash reward",
        "Congratulations you are a lucky winner",
        "Get a special discount offer today",
        "Earn money quickly from home",
        "Limited time offer buy now",
        "Your bank statement is ready",
        "Please send me the project report",
        "The lecture will begin at nine",
        "Your assignment submission is tomorrow",
        "Let us meet in the library",
        "The team meeting is scheduled for Monday",
        "Please review the attached document",
        "Your exam timetable has been published",
        "Thank you for attending the meeting",
        "Win a brand new smartphone today"
    ],
    "label": ["spam","spam","spam","spam","spam","spam","ham","ham","ham","ham","ham","ham","ham","ham","ham","spam"]
}
df = pd.DataFrame(data)
display(df)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df["email"], df["label"], test_size=0.25, random_state=42, stratify=df["label"]
)

model = Pipeline([
    ("vectorizer", CountVectorizer()),
    ("classifier", MultinomialNB())
])
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Accuracy:", round(accuracy_score(y_test, y_pred) * 100, 2), "%")
print("\nClassification report:")
print(classification_report(
    y_test, y_pred, labels=["ham", "spam"],
    target_names=["Ham (Not Spam)", "Spam"], zero_division=0
))
cm = confusion_matrix(y_test, y_pred, labels=["ham", "spam"])
print("\nConfusion matrix:")
display(pd.DataFrame(cm, index=["Actual Ham", "Actual Spam"], columns=["Predicted Ham", "Predicted Spam"]))
```

```python
new_email = ["Congratulations! You have won a free smartphone"]
prediction = model.predict(new_email)[0]
print("New email:", new_email[0])
print("Prediction:", prediction)
```
